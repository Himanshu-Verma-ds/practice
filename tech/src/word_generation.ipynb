```python
import string 
import itertools
import pandas as pd
```

```python
all_words= [''.join(p) for p in itertools.product(string.ascii_lowercase, repeat= 5)]
len(all_words)
```

```python
pattern = ['u', 'n', '*', '*', 'd']
letters = string.ascii_lowercase

def generate_patterned_words(pattern):
    slots = [letters if ch == '*' else [ch] for ch in pattern]
    return [''.join(p) for p in itertools.product(*slots)]

words = generate_patterned_words(pattern)
print(pd.DataFrame(words))
print("Total:", len(words))
```

#### 1. Abstract Method and class
- we use it to enforce a blueprint
- Any class not following that, it's object won't be created
- Maintains consistency and polymorphism
- Ex- All the different modes of payments must have 'pay' and 'refund' method

```python
from abc import ABC, abstractmethod

# here the PaymentMethod class serves as blueprint to other classes
class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass 
    
    @abstractmethod
    def refund(self, amount):
        pass
    

class CreditCardPayment(PaymentMethod):
    def pay(self, amount):
        print(f'Amount {amount} paid using credit card !!')
    
    def refund(self, amount):
        print(f'Amount {amount} refunded which was paid using credit card !!')
        
    
class UPIPayment(PaymentMethod):
    def pay(self, amount):
        print(f'Amount {amount} paid using UPI !!')
    
    def refund(self, amount):
        print(f'Amount {amount} refunded which was paid using UPI !!')
        

class DebitCardPayment(PaymentMethod):
    def pay(self, amount):
        print(f'Amount {amount} paid using debit card !!')
    
```

```python
credit= CreditCardPayment()
print(credit.pay(1000))
print(credit.refund(1000))
```

```python
upi= UPIPayment()
print(upi.pay(1000))
print(upi.refund(1000))
```

```python
# here as DebitCardPayment does not contain 'refund' method which was specified in BluePrint 'PaymentMethod' class it gave error
# Even the object was not created !!

debit= DebitCardPayment()
debit
```

#### 2. PolyMorphism
- It is achieved when all related classes have the same method name serving the same role, but with different implementations
- Each class should implement the same method name (and ideally the same method signature)
- That method should serve the same purpose (e.g., pay(amount) always processes a payment)
- But the implementation can differ depending on the class

```python
class CreditCardPayment:
    def pay(self, amount):
        print(f"Paid {amount} using Credit Card !!")

class PayPalPayment:
    def pay(self, amount):
        print(f"Paid {amount} using PayPal !!")

class UPIPayment:
    def pay(self, amount):
        print(f"Paid {amount} using UPI !!")
```

```python
for payment in [CreditCardPayment(), PayPalPayment(), UPIPayment()]:
    payment.pay(500)
```

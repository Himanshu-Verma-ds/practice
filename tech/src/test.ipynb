```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity


model= SentenceTransformer('all-MiniLM-L6-v2')

ground_truth_text= 'The philosopher spoke of a river that never rests, where each drop replaces another, yet the river itself persists unchanged. He claimed this was how identities endure despite the constant erosion of time'
candidate_text= 'In the study of personal memory, researchers note that although neurons are replaced and synapses rewired, the continuity of consciousness is preserved — much like a flowing current that retains its name while its waters move on'

ground_truth_embeddings = model.encode(ground_truth_text, convert_to_tensor=True , 
                                       show_progress_bar=True)

candidate_embeddings = model.encode(candidate_text, convert_to_tensor=True , 
                                    show_progress_bar=True)

cosine_scores = cosine_similarity(ground_truth_embeddings, candidate_embeddings)
cosine_scores
```

##### Questions:

1. Why the real world data is not completely linear?
- Bcz there are real world factors affecting it (called stochastic errors) which cannot be modelled or captured, basically cannnot be explained by the factors [Ex- A guy with low cgpa, getting higher package maybe bcz his interview went really well or there was some referral or other network]


2. Intuition of 'm' and 'b'
- 'm' defines the impact (weightage) of a particular independent variable on dependent variable
- 'b' is the offset, allows model to have real world thing, if exp is 0, it does not mean salary is 0

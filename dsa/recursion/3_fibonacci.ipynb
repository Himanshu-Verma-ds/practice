#### Write a recursive function to give fibonacci of nth number
- 0, 1, 1, 2, 3, 5, 8...
- 0th fibonacci: 0
- 1st fibonacci: 1
- 2nd fibonacci: 1

### Recursion Tree



### Remember 

- Here in this recursion tree, first we travel to the extreme left branch, then level wise we move up
- So the left calls are all executed first and then the right ones
- So the ones written before are executed first (last+slast), here the entire 'last' calls are executed first


#### TC:
- For every call you are making 2 recursion calls, f(n) calls f(n-1) and f(n-2)
- So it is exponential TC, 2^n calls

```python
def fibonacci_recursion(n: int):
    
    # base case
    if n <= 1: # if n=0, returns 0, if n=1 returns 1
        return n
    
    last= fibonacci_recursion(n-1) 
    slast= fibonacci_recursion(n-2)
    
    return last+slast
```

```python
fibonacci_recursion(n=3)
```

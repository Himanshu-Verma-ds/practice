#### Problem Statement:

- Return the sum of all the possible subsequences in ascending order

```python
def subset_sum(arr: list,
               idx: int,
               curr_sum: int,
               curr: list= None,
               ans: list= None):
    
    if curr is None:
        curr= []
    
    if ans is None:
        ans= []
    
    # base case
    if (idx == len(arr)):
        ans.append(curr_sum)        # or ans.append(sum(curr))
        return 

    
    # Pick case
    curr.append(arr[idx])
    curr_sum += arr[idx]
    subset_sum(arr, idx+1, curr_sum, curr, ans)
    
    # backtrack (undo the changes)
    curr.pop()
    curr_sum -= arr[idx]
    subset_sum(arr, idx+1, curr_sum, curr, ans)
    
    # Sort the ans array, containing the sum of the elements
    ans.sort()
    

    return ans
```

```python
arr= [3,1,2]
ans= subset_sum(arr= arr, 
                idx= 0, 
                curr_sum= 0)

ans
```

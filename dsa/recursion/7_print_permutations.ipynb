#### Problem Statement:

- Generate all the possible permutations of a given array
- No. of possible permutations: n!


Approach:
- Keep a visited array which keeps a track of the elements visited so far
- Loop through the entire array elements

```python
def print_permutations(arr, ans=None, visited=None):
    n = len(arr)
    
    if visited is None:   # initialize only once
        visited = [False] * n
        ans = []

    # base case
    if len(ans) == n:
        print(ans)
        return

    # loop through the entire array
    for i in range(n):
        if not visited[i]:
            visited[i] = True
            ans.append(arr[i])

            print_permutations(arr, ans, visited)

            ans.pop()
            visited[i] = False


# Example
print_permutations([1, 2, 3])
```

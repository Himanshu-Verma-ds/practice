


#### Problem Statement:
- Generate the subsequences where the sum of elements in it is equal to the target
- You can pick any element any number of times


#### Approach:
- Base case will be when idx reaches the end of array and the target becomes zero
- If you decide to pick an element multiple times, you still stay at same idx
- You should be picking an element only if element to be picked is less than target
- Target should reduce by the amount of new element added
- For the not pick case, move to next index, target remains same and element added to curr_array should be removed

```python
def combination_sum(arr: list,
                    idx: int,
                    curr: list,
                    target: int,
                    ans: list):
    
    if curr is None:
        curr = []
        
    if ans is None: 
        ans= []
        
    # base case
    if (idx == len(arr)):           
        if (target == 0):
            ans.append(curr[:])     # add a copy of the 'curr' array
        return
    
    # case to add an element multiple times
    if (target >= arr[idx]):        # pick the element only if the element is less or equal to the target
        curr.append(arr[idx])
        combination_sum(arr, idx, curr, target-arr[idx], ans)   # if element is picked then the target reduces by the amount of the element
    
        # backtrack (undo the changes, but only if the element is added [pick case])
        curr.pop()
    
    # not pick case (hence move the idx forward)
    combination_sum(arr, idx+1, curr, target, ans)
    
    return ans
```

```python
arr= [2,3,6,7]

ans= combination_sum(arr= arr,
                     idx= 0,
                     curr= None,
                     target= 7,
                     ans= None)
ans
```

```python
# Function to print name 'n' times using recursion

def print_name(count: int,
               total: int):
    
    # base case
    if (count > total):     # when base case is reached the function terminates
        return 
    
    print('Himanshu')
    print_name(count+1, total) # increment the counter value
```

```python
total= int(input('Enter the number of times to print name: '))
print_name(count=1, total= total)
```

#### Space and Time complexity for recursion:

- SC:
    - Consider the function calls waiting in stack space
    - Do not consider the function call of the base case
    

```python
# Function to print linearly from 1 to n

def print_n(count: int,
            n: int):
    
    # base case
    if(count > n):
        return
    
    print(count)
    print_n(count+1, n)
```

```python
n= int(input("Enter the number till which to be printed: "))
print_n(count=1,
        n= n)
```

#### Function to print linearly from 1 to n, but NOT USE i+1 in function argument
- Meaning we can use i-1, thereforewe can write print after function call



```python
# Function to print linearly from 1 to n, but NOT USE i+1 in function argument
# Meaning we can use i-1, thereforewe can write print after function call

def print_without_increment(count: int,
                            n: int):
    
    # base case
    if (count < 1):
        return
    
    print_without_increment(count-1, n)
    print(count)
```

```python
n= 5
print_without_increment(count= n,
                        n= n)
```

#### 1. Sum of n numbers

```python
# Sum of n numbers

def sum_numbers(n: int):
    
    # base case
    if n == 1:      # as sum upto 1 is 1
        return 1
    
    # if not base case then make another function call
    return n + sum_numbers(n-1)
```

```python
n= int(input("Enter the number upto which sum is to be found: "))
print(sum_numbers(n))
```

#### 2. Factorial using recursion

```python
# Factorical upto n numbers

def factorial(n: int):
    
    # base case
    if n == 1:
        return 1    # as facorial(1)= 1
    
    # recursive case
    return n*factorial(n-1)
```

```python
n= int(input("Enter the number for which factorial is to be found: "))
print(factorial(n))
```

#### 3. Reverse an array using recursion

```python
def reverse_array(arr: list,
                  left: int,
                  right: int):
    
    # base case
    if (left >= right):
        return 
    
    # recursive call
    arr[left], arr[right]= arr[right], arr[left]
    reverse_array(arr, left+1, right-1)
```

```python
arr= [1,2,3,4,5]
reverse_array(arr= arr,
            left= 0,
            right= len(arr)-1)
print(arr)
```

#### 4. Check Pallindrome

```python
def check_pallindrome(arr: list, 
                      left: int,
                      right: int):
    
    # base case
    if arr[left] != arr[right]:
        return False
    
    # another base case (if string is palindrom above base case will never get trigerred, hence we stop once pointers cross each other)
    if (left >= right):
        return True # meaning the above base case was never triggered, therefore all the elements were equal
    
    # Recursive call (if the base case was not reached, check for other places or elements)
    return check_pallindrome(arr, left+1, right-1)
```

```python
arr= "123212"
check_pallindrome(arr= arr,
                  left= 0, 
                  right= len(arr)-1)
```

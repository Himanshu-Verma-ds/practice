#### Constant window

Given 'k' sized array you have to find the one with maximum sum

- Start with 2 pointers representing the length of constant window
- Run till the right pointer does not cross the last index
- Calculate the sum, remove the left pointer from sum, add sum of right pointer
- Increment both left and right pointer, to slide the window

```python
def constant_sliding_window(arr: list, 
                            window: int):
    
    left= 0
    right= window-1
    
    curr_sum= sum(arr[left: right+1])
    max_sum= curr_sum  # might be the case that the first curr_sum is the best one itself
    
    while(right < len(arr)-1):
        curr_sum = curr_sum - arr[left]
        left += 1
        
        right += 1
        curr_sum = curr_sum + arr[right]
        
        max_sum= max(curr_sum, max_sum)
    return max_sum
```

```python
arr= [-1, 2, 3, 3, 4, 5, -1]

max_sum= constant_sliding_window(arr, 4)
max_sum
```

#### Flexible window

Find the maximum window length with sum <= k

- Start with both pointers at same element, meaning the length of window=1
- Expand the window till the sum of the window is <= target
- If not shrink the window by removing the left element from sum and moving left pointer ahead

```python
def longest_subarray_with_sum_k(arr: list,
                                target: int):
    
    left= 0
    right= 0
    curr_sum= 0
    max_len= float('-inf')
    
    while(right < len(arr)):
        
        # calculate the sum of the window subarray
        curr_sum += arr[right]
        
        # Remove the left element from the windowed subrray until the curr_sum of the window <= target
        while(curr_sum > target): # basically shrinking the window until condition gets matched
            curr_sum -= arr[left]
            left += 1
        
        # When the curr_sum <= target, then update the max_len (do the comparison)
        if (curr_sum <= target):
            max_len= max(max_len, left-right+1)
        
        right += 1
    
    return max_len
```

#### Problem:
- Given an array depicting the type of fruits, there are 2 baskets each can hold only 1 type of fruit
- Find the maximum number of fruits you can pick and fill the baskets
- Keeping one basket entirely empty is also allowed

#### Brute Force:
- Generate all the subrrays
- Keep a set and everytime add the new element in set
- Whenever length of set crosses threshold (here 2), break off
- Otherwise add in the ans list, or compare the max_len with current len of generated subarray

```python
def fruits_in_basket_brute(arr: list):
    
    max_len= 0
    ans= []
    for i in range(len(arr)):
        s= set()
        for j in range(i, len(arr)):
            s.add(arr[j])
            if len(s) > 2:
                break
            
            ans.append(arr[i: j+1])
            max_len= max(max_len, j-i+1)
    print(ans)
    return max_len
    
```

```python
arr= [3,3,3,1,2,1,1,2,3,3,4]
ans= fruits_in_basket_brute(arr)
ans
```

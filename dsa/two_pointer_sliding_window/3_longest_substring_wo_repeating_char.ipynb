```python
def longest_substring_brute(string: str):
    max_len= 0

    for i in range(len(string)):
        hashmap= {}
        for j in range(i, len(string)):
            if(string[j] in hashmap):
                break        # break away as that element has already been seen, so no point considering that as substring
            
            
            hashmap[string[j]]= 1   # mark new character as seen, so that if it occurs next time substring breaks
            length= j-i+1
            max_len= max(max_len, length)
    
    return max_len
```

```python
string= "cadbzabcd"

max_len= longest_substring_brute(string)
max_len
```

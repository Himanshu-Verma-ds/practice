#### Collect max points from cards, picking only k either from start or end contiguously

- Maintain left_sum and right_sum, max_sum will be left_sum + right_sum
- Use all the k elements from left to make up left sum
- Now start from right and remove 1 element from left_sum and add 1 to right_sum

```python
def collect_max_points(arr: list,
                       k: int):
    
    left_sum= 0
    right_sum= 0
    max_sum= float('-inf')
    
    # make up the left sum
    for i in range(k):
        left_sum += arr[i]
    
    
    # start from window size-1, upto 0 and remove 1 element from left sum, add 1 to right sum
    r_index= len(arr)-1  # this keeps track of right_sum, increasing it every time
    for i in range(k-1, 0, -1):
        left_sum -= arr[i]
        right_sum += arr[r_index]
        
        max_sum= max(max_sum, left_sum+right_sum)
        r_index -= 1
    
    return max_sum
```

```python
arr= [6,2,3,4,7,2,1,7,1]
k= 4

max_sum= collect_max_points(arr, k)
max_sum
```

#### Left rotate an array by 'd' places

##### Brute force 
1. store 'd' elements in another array (temp) 
2. shift elements after d to n to starting of org array
3. fill the org array left places with temp array elements

Space complexity= O(d) 
Time complexity= O(d) + O(n-d) + O(d) 

```python
arr= [1,2,3,4,5,6]
d= 3

temp=[]
for i in range(d):
    temp.append(arr[i])
    
for i in range(1, d+1):
    arr[i-1]= arr[d+i+1]

arr
```

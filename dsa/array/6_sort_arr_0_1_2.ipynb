##### Brute Force
- Just use sorting function

#### Better Approach
- Keep 3 counters, one for each element
- Then manually insert those elements in array

```python
def sort_zero_one_two_better(arr: list):
    
    count_zero= 0
    count_one= 0
    count_two= 0
    
    for i in range(len(arr)):
        if arr[i]== 0:
            count_zero += 1
        
        elif arr[i] == 1:
            count_one += 1
        
        else:
            count_two += 1
    
    # Fill zeros
    for i in range(count_zero):
        arr[i]= 0
    
    # Fill ones  
    for i in range(count_zero, count_zero+count_one):
        arr[i]= 1
    
    # Fill twos    
    for i in range(count_zero+count_one, count_zero + count_one + count_two):
        arr[i]= 2
        
    return arr
```

```python
arr= [0,1,2,0,1,2,1,2,0,0,0,1]

ans= sort_zero_one_two_better(arr)
ans
```

#### Dutch National Flag Algo

Intuition:
- 0 to (low-1): All 0s
- low to (mid-1): All 1s
- mid to high: unsorted (0/1/2) ---> consider the array to sort fall in this zone
- (high+1) to (n-1): All 2s

Algo:
- If arr[mid] == 0, then swap (arr[mid], arr[low]), mid++, low--
- If arr[mid] == 1, already sorted part is there so just move mid++
- If arr[mid] == 2, swap (arr[mid], arr[high]), high--
- Now whatever is element at arr[mid], after 3rd case that will be dealt accordingly


TC:
- O(n), as you are looping through all the elements in array

SC:
- O(1), as you are not taking any extra space

```python
def sort_zero_one_two_optimal(arr: list):
    low=0
    mid=0
    high= len(arr)-1
    
    while(mid <= high):
        if arr[mid] == 0:
            arr[mid], arr[low]= arr[low], arr[mid]
            mid += 1
            low += 1
        
        if arr[mid] == 1:
            mid += 1
        
        if arr[mid] == 2:
            arr[mid], arr[high]= arr[high], arr[mid]
            high -= 1
    
    return arr
```

```python
arr= [0,1,2,0,1,2,1,2,0,0,0,1]

ans= sort_zero_one_two_optimal(arr)
ans
```

#### Brute force

1. store the last element in tmp variable
2. store the values of next element at prev places in array
3. place the tmp variable value at the end of array

```python
# Rotate the array by 1 place
def rotate_arr(arr):
    tmp= arr[0]
    
    for i in range(1, len(arr)):
        arr[i-1]= arr[i]
    
    arr[-1]= tmp
    return arr

# Example
arr= [1,2,3,4,5]
print(rotate_arr(arr))
```

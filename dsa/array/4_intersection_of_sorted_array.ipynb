#### Brute Force:

- Traverse both arrays
- Take visited array with size equal to 2nd array
- If there is a match and visited array is 0 for that index take that element
- Visited array is taken as there can be duplicate elements in ans array

```python
def intersect_sorted_array_brute(arr1: list,
                                 arr2: list):
    
    
    vis= [0]*len(arr2)
    ans= []
    
    for i in range(len(arr1)):
        for j in range(len(arr2)):
            if (arr1[i] == arr2[j] and vis[j] == 0):
                ans.append(arr1[i])
                vis[j] = 1
                break
                
            if (arr2[j] > arr1[i]):
                break
    
    return ans


# Example
arr1= [1,1,2,3,4,5]
arr2= [2,3,4,4,5,6]
print(intersect_sorted_array_brute(arr1, arr2))
```

#### Problem:
- There will be an even length array with half elements as positive and half negative
- Task is to assign them with alternate signs (+,-,+,-,+,-)

#### Brute Force:

- Take positive array, store all positive elements here
- Take negative array, store all negative there
- Run another n/2 loop, every 2*i index has positive array elements
- 2*i+1 has all negative elements


- SC: O(2N), for pos and neg list
- TC: O(N) + O (N/2)

```python
def rearrange_elements_brute(arr: list):
    
    pos= []
    neg= []
    for i in range(len(arr)):
        if arr[i] < 0:              # if element is negative, store it in negative array
            neg.append(arr[i])
        
        else:
            pos.append(arr[i])      # if element is positive, store it in positive array
    
    for i in range(len(arr)//2):    # run loop to n/2 length
        arr[2*i]= pos[i]            # fill even indices with positive arr elements
        arr[2*i+1]= neg[i]          # fill odd indices with negative arr elements            
        
    return arr
```

```python
arr= [3, 1, -2, -5, 2, -4]
ans= rearrange_elements_brute(arr)
ans
```

#### Doing it in 1 single pass

```python
def rearrange_elements_by_sign(arr: list):
    pos= 0
    neg= 1
    ans= [float('-inf')]*len(arr)
    
    for i in range(len(arr)):
        if arr[i] > 0:
            ans[pos]= arr[i]
            pos+= 2
        
        else:
            ans[neg]= arr[i]
            neg += 2
    
    return ans
```

```python
arr= [3, 1, -2, -5, 2, -4]
ans= rearrange_elements_by_sign(arr)
ans
```

#### Variety 2: When there are unequal no. of pos and neg numbers

- Fall back to the brute force solution
- Storing pos and neg in different lists/arrays
- If pos elements > negative elements
    - Loop till negative elements list, fill the elements
    - Store idx of 2*len(neg), as this is where till array has been filled
    - Now, loop from idx to len(pos), storing the pos elements at idx and incrementing idx

- If neg elements > pos:
    - Do the same as above for that case

#### TC:
- Loop to store pos/neg elements + Loop upto min(pos, neg) + Loop till remaining elements
- O(n) + O(0) [only pos/neg elements] + O(n) [case when only neg/pos elements are there]
- O(n) + O(n/2) [half pos, half neg elements] + O(0) [case when half neg and half pos elements are there]

```python
def rearrange_unequal_elements_by_sign(arr: list):
    
    pos= []
    neg= []
    
    for i in range(len(arr)):
        if arr[i] < 0:
            neg.append(arr[i])
        
        else:
            pos.append(arr[i])
    
    # Checking which one has less element (pos array has more elements) 
    if len(pos) > len(neg):
        for i in range(len(neg)):   # till they are equal fill them like previous case by going till len of whichever list has less elements (here neg list)
            arr[2*i]= pos[i]
            arr[2*i+1]= neg[i]
        
        idx= len(neg)*2     # finding the index from where to start filling the remaining elements
        for i in range(idx, len(pos)):  # start from that index to len of longer array
            arr[idx]= pos[i]    # fill that idx of org arr with longer array elements
            idx += 1
    
    # neg array has more elements
    else:
        for i in range(len(pos)):
            arr[2*i]= pos[i]
            arr[2*i+1]= neg[i]
        
        # fill the remaining neg array elements
        idx= len(pos)*2
        for i in range(idx, len(neg)):
            arr[idx]= neg[i]
            idx += 1
    
    # return the org array
    return arr 
```

```python
arr= [3, 1, -2, -5, 2, -4, -6, -8, -12]
ans= rearrange_unequal_elements_by_sign(arr)
ans
```

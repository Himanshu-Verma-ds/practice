#### Brute Force:

<font size= '2.5'>
- Take a set and convert it into list

Space complexity: 
- Set takes: 
    - set(m) -> O(m) 
    - set(n) -> O(n)
    - Total: O(m+n)
- List takes: O(m+n)
- So finally: O(2m+2n) -> O(m+n)


Time Complexity:
- same as each operation takes constant time, so m operation takes O(m) time
- set(arr1): O(n)
- set(arr2): O(m)
- union: O(n + m)
- list conversion: O(n + m)
- **Total: O(n + m)**

</font>



```python
def union_of_sorted_arrays_brute(arr1: list,
                                 arr2: list):
    
    return list(set(arr1+ arr2))


# example

arr1= [1,1,2,3,4,5]
arr2= [2,3,4,4,5,6]
print(union_of_sorted_arrays_brute(arr1, arr2))
```

#### Optimal approach:

- Take 2 pointers place them at starting of 2 arrays
- Traverse till both of them are <= len of respective arrays
- When one array gets exhausted fill the elements of other array in ans array
- <font size= '2'> Video: https://youtu.be/wvcQg43_V8U?si=copVya70tp4Zc-Dk&t=3005 </font>

- Time Complexity:
    - 1 iteration in array of size m, another in array of size n
    - Total: O(m+n)

- Space Complexity:
    - For returning the answer and not for solving it O(m+n)

```python
def union_of_sorted_arrays_optimal(arr1: list,
                                   arr2: list):
    
    i= 0
    j= 0
    ans= []
    
    # when both indices are well within size of arrays
    while(i < len(arr1) and j < len(arr2)):
        if arr1[i] <= arr2[j]:
            if (len(ans) == 0 or ans[-1] != arr1[i]):
                ans.append(arr1[i])
            
            i+= 1  # move i pointer in any case (do dry run)
        
        else:
            if (len(ans) == 0 or ans[-1] != arr2[j]):
                ans.append(arr2[j])
            
            j+= 1
                
    
    # if arr2 gets exhausted, fill ans array with elements of arr1
    while(i < len(arr1)):
        if (len(ans) == 0 or ans[-1] != arr1[i]):
            ans.append(arr1[i])
        
        i+= 1
            
            
    # if arr1 gets exhausted, fill ans arrays with elements of arr2
    while(j < len(arr2)):
        if (len(ans) == 0 or ans[-1] != arr2[j]):
            ans.append(arr2[j])
            
        
        j += 1
    
    return ans


# example
arr1= [1,1,2,3,4,5]
arr2= [2,3,4,4,5,6]
print(union_of_sorted_arrays_optimal(arr1, arr2))
```

#### Brute Force

- store the non-zeros in new array whenever found while traversing the org array
- fill the entire array with non-zero temp array elements and remaining array with 0

Space complexity: O(n)
Time complexity: O(n) + O(n)

```python
def move_zeros_to_right_brute(arr: list):
    
    temp= []    # will store non-zero elements
    for i in range(len(arr)):
        if arr[i] != 0:
            temp.append(arr[i])
    
    # now fill the org array with elements of temp (non-zero array)
    for i in range(len(temp)):
        arr[i]= temp[i]
        
    
    # now fill the remaining places starting from length of temp to lenght of org array with 0
    for i in range(len(temp), len(arr)):
        arr[i]= 0
    
    return arr


# Example
arr= [1,0,0,3,5,0,7,0,8]
print(move_zeros_to_right_brute(arr))
```

#### Optimal approach:

- Take a pointer (j) put it at the 1st occured zero
- Start another pointer (i), 1 ahead the first pointer (j)
- Whenever i points at non-zero, swap and move both pointer, when i encounters 0, no swapping just move i pointer
- Do it till i crosses the lenght of array

```python
def move_zeros_to_right_optimal(arr: list):
    
    # find the first occurence of 0
    j= -float('inf')
    for i in range(len(arr)):
        if arr[i] == 0:
            j=i
            break
    
    # start i from 1 index ahead of j
    for i in range(j+1, len(arr)):
        if arr[i] != 0:
            # swap when i encounters non-zero value
            arr[i], arr[j]= arr[j], arr[i]
            j += 1
    
    return arr


# Example
arr= [1,0,0,3,5,0,7,0,8]
print(move_zeros_to_right_optimal(arr))
            
```

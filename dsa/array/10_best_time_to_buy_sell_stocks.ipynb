#### Problem Statement:
- You are given stock prices (arr) where arr[i] is the price of the stock on day i.
- You want to maximize profit by choosing one day to buy and one later day to sell.
- You cannot sell before buying.

#### Approach:
- For the 1st iteration assume, you bought on day 1 and sold it on same day so profit is zero
- mini: keeps track of the minimum stock price seen so far (best day to buy)
- cost: the profit you would make if you buy at mini and sell today (i-th day)
- profit: Max profit found so far

#### Intuition:

- At each day:
    - Assume we buy at mini and sell today → that’s cost.
    - Update profit if this cost is larger than previous best.
    - Update mini if today’s price is lower (because maybe we should buy today instead)

```python
def best_time_to_buy_sell_stock(arr: list):
    mini= arr[0]    # minimum price seen so far, for 1st day, it will be that day price only
    profit= 0       # for 1st day, the profit will be zero, as the mini value before that day does not exist 

    for i in range(1, len(arr)):
        cost= arr[i] - mini     # profit made if I buy on price of mini and sell it on i-th day
        profit= max(profit, cost)   # max profit seen so far
        mini= min(mini, arr[i])     # update mini to minimum price seen so far
    return profit
```

```python
arr= [7,1,5,3,6,4]
profit= best_time_to_buy_sell_stock(arr)
profit
```

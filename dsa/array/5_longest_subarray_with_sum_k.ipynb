```python
# generating all the subarrays (they are contiguous)

arr= [1,2,3,4,5,6]
target= 5

ans=[]
for i in range(len(arr)):
    for j in range(i, len(arr)):
        ans.append(arr[i:j+1])
        
print('ans: ', ans)

final_arr= []
for i in ans:
    if sum(i) == target:
        final_arr.append(i)

print('final_arr: ', final_arr)
```

#### Brute Force:
1. Generate all the subequences
2. Add only the new element coming up
3. Store the maximum length of subarray in a variable

```python
def longest_subarray_with_sum_k_brute(arr: list,
                                      k: int):
    
    max_len= 0
    for i in range(len(arr)):
        s= 0
        for j in range(i, len(arr)):
            s += arr[j]  # only add the new element to the sum
            
            if s == k: # if the sum of elements in subarray matches the target (k)
                max_len= max(max_len, j-i+1) # update the max length of the potential subarray
    return max_len
```

```python
arr= [1,2,3,4,5,6]
target= 5

max_len= longest_subarray_with_sum_k_brute(arr, target)
max_len
```

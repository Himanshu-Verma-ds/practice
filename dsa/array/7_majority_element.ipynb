#### Find out the element which is occuring > Floor(n/2)

#### Better Approach:

TC:
- O(n) [array iteration to build hashmap] + O(n) [hashmap iteration]= O(2n)

SC:
- O(n): Extra space taken by hashmap (dict)

```python
def majority_element_brute(arr: list):
    
    # hash all the element in dict
    d= {}
    for i in range(len(arr)):
        if arr[i] not in d:
            d[arr[i]]= 1
        
        else:
            d[arr[i]] += 1
    
    # iterate in dict and find the element having count > n/2
    for num, count in d.items():
        if count > len(arr)/2:
            return num 
    
    return 'None'
```

```python
arr= [2,2,3,3,1,2,2]

ans= majority_element_brute(arr)
ans
```

#### Moore's Voting Algo

- Start with 1st element assuming it to be majority one
- Increase counter everytime you see that element
- Whenever you see any element other than the one you assumed to be highest time occuring, decrement counter
- Whenever counter reaches zero consider other element as most occuring and repeat
- Intuition: If the counter > n/2 of element it won't get cancelled by count of other elements

- At last, verify if the element you considered is really the highest occuring by iterating through array again

```python
def majority_element_optimal(arr: list):
    
    count= 0
    for i in range(len(arr)):
        if count == 0: # if count reaches 0, consider new element as the one having highest occurence
            element= arr[i]
            count= 1
        
        elif arr[i] == element:  # if the element in array is one we considered increase counter
            count += 1
        
        else:       # if not then decrease counter
            count -= 1
    
    # verify the element we got from Moore's voting algo is actually the one having highest occurence
    check_counter=0
    for i in range(len(arr)):
        if arr[i] == element:
            check_counter += 1
    
    if check_counter > len(arr)/2:
        return element
    
    return 'None'
```

```python
arr= [2,2,3,3,1,2,2]

ans= majority_element_optimal(arr)
ans
```

#### Brute Force
- Generate all the subarrays
- Find the one with maximum sum

```python
def max_subarray_brute(arr: list):
    
    # Generated and stored all the subarrays
    ans= []
    for i in range(len(arr)):
        for j in range(i, len(arr)):
            ans.append(arr[i:j+1])
            
    # Check for the one having maximum sum
    max_sum= float('-inf')
    max_subarray= []
    
    for subarray in ans:
        if sum(subarray) > max_sum:
            max_sum= sum(subarray)
            max_subarray= subarray
            
    return max_subarray, max_sum       
```

```python
arr= [-2, -3, 4, -1, -2, 1, 5, -3]
max_subarray, max_sum= max_subarray_brute(arr)

print('Max subarray: ', max_subarray)
print('Max sum: ', max_sum)
```

#### Second brute force

```python
def max_subarray_brute_2(arr: list):
    
    # Generated and stored all the subarrays
    max_sum= float('-inf')
    for i in range(len(arr)):
        sum_= 0
        for j in range(i, len(arr)):
            
            # update the subarray sum everytime by adding the elements
            sum_ += arr[j]
            max_sum = max(max_sum, sum_) # compare the subarray sum with global max_sum
            
    return max_sum
```

```python
arr= [-2, -3, 4, -1, -2, 1, 5, -3]
max_sum= max_subarray_brute_2(arr)

print('Max sum: ', max_sum)
```

#### Kadane's Algo (find contiguous subarray with largest possible sum)

- At every step, you want to add current element to the running sum.
- If the running sum becomes negative, you reset it to 0 (since a negative prefix can’t help future subarrays)

```python
def kadane(arr: list):
    
    max_sum= float('-inf')
    curr_sum= 0
    
    for i in range(len(arr)):
        curr_sum += arr[i]      # add the elements to sum
        max_sum= max(curr_sum, max_sum) # update the max_sum
        
        if curr_sum < 0: # reset sum to 0 if it become negative as it does not makes sense to carry negative sum, that will decrease the sum only
            curr_sum = 0
    
    return max_sum
```

```python
arr= [-2, -3, 4, -1, -2, 1, 5, -3]
max_sum= kadane(arr)

print('Max sum: ', max_sum)
```

```python
def max_subarray_sum_brute(arr: list):
    
    # generate all the subarrays
    ans= []
    for i in range(len(arr)):
        for j in range(i, len(arr)):
            ans.append(arr[i:j+1])

    # iterate through all the generated subarrays and find the one with largest sum
    maxi= float('-inf')
    max_subarray= []
    for sub_array in ans:
        if sum(sub_array) > maxi:
            maxi= sum(sub_array)
            max_subarray= sub_array
    return max_subarray, maxi
```

```python
arr= [-2,-3,4,-1,-2,1,5,-3]
max_subarray, maxi= max_subarray_sum_brute(arr)
print(max_subarray)
print(maxi)
```

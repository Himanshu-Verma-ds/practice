```python
def max_subarray_sum_brute(arr: list):
    
    # generate all the subarrays
    ans= []
    for i in range(len(arr)):
        for j in range(i, len(arr)):
            ans.append(arr[i:j+1])

    # iterate through all the generated subarrays and find the one with largest sum
    maxi= float('-inf')
    max_subarray= []
    for sub_array in ans:
        if sum(sub_array) > maxi:
            maxi= sum(sub_array)
            max_subarray= sub_array
    return max_subarray, maxi
```

```python
arr= [-2,-3,4,-1,-2,1,5,-3]
max_subarray, maxi= max_subarray_sum_brute(arr)
print(max_subarray)
print(maxi)
```

#### Brute Force
- Loop through all the elements in an array and check if they add up to target
- If yes return the indices

```python
def two_sum_brute(arr: list,
                  target: int):
    
    for i in range(len(arr)):
        for j in range(len(arr)):
            if i == j:
                continue
            
            if arr[i] + arr[j] == target:
                return (i,j)
    
    return "None"
```

```python
arr= [2,6,5,8,11]
target= 14

ans= two_sum_brute(arr,
                  target)

ans
```

#### Better Approach
- Loop through array
- Calculate the more element required to reach target
- If present in dictionary then return the indices
- If not then add the number (not complement) to dict

##### <u>Time Complexity </u>

- We loop through the list once → n iterations (where n = len(nums)).
- Inside the loop, complement in seen → dictionary lookup → O(1) on average (hash table).
- seen[num] = i → dictionary insertion → O(1) on average.
- Total time complexity = O(n)

##### <u>Space Complexity</u>

- We store each number (at most once) in the dictionary seen.
- Each entry stores, key = number, value = index → constant size per entry.
- In the worst case (when no two numbers sum to target), the dictionary stores all n elements.
- Total space complexity = O(n)

```python
def two_sum_better(arr: list,
                   target: int):
    d= {}
    for idx, num in enumerate(arr):
        complement= target-num # how much more is required to reach target
        if complement in d:
            return (d[complement], idx) # if complement already present in dict, return the indices
        
        d[num]= idx # if complement not present then add the 'num', with key as num and value as idx
    return 'not present'
```

```python
arr= [2,6,5,8,11]
target= 14

ans= two_sum_better(arr,
                  target)
ans
```

#### Optimal Approach (only when only yes/no is asked and not the indices)

- Sort the array
- 2 pointer
- If sum, less than target increment left pointer (as from left to right the value increases)
- If sum, more than target decrement right pointer 

```python
def two_sum_optimal(arr: list,
                    target: int):
    
    arr= sorted(arr)
    left= 0
    right= len(arr)-1
    
    while(left<right):
        summ= arr[left]+arr[right]
        
        if summ == target:
            return 'yes'
        
        elif (summ < target):
            left+= 1
        
        else:
            right -= 1

    return 'not present'
```

```python
arr= [2,6,5,8,11]
target= 14

ans= two_sum_optimal(arr,
                  target)

ans
```

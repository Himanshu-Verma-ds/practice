```python
# Find the largest element in an array

def find_largest_element(arr: list) -> int:
    
    largest= arr[0]
    for i in range(1, len(arr)):
        if arr[i] > largest:
            largest= arr[i]
    
    return largest
    

# function call
arr= [3,2,1,5,2,8]
largest= find_largest_element(arr)
print(largest)
```

#### Find the second largest element in an array

```python
"""
Brute Approach: 
1. Sort the array O(nlogn)
2. Take last element as the largest one
3. Start from second last index and find the element that is not equal to the largest and BREAK, that will be the second largest element

Time comlexity: O(nlogn) + O(n)= O(nlogn)
"""

def find_second_largest_element_brute(arr: list) -> tuple:
    
    sorted_arr= sorted(arr)
    largest= sorted_arr[-1]
    
    sec_largest= float('-inf')
    for i in range(len(sorted_arr)-2, -1, -1):
        if sorted_arr[i] != largest:
            sec_largest= sorted_arr[i]
            break
    
    return largest, sec_largest


# function call
arr= [3,2,1,5,2,8]
largest, sec_largest= find_second_largest_element_brute(arr)
print(largest, sec_largest)
```

```python
"""
Better Approach: 
1. In 1st pass find the largest element: O(n)
2. In 2nd pass find the element bigger than rest but smaller than largest: O(n)

Time comlexity: O(n) + O(n)= O(n)
"""

def find_second_largest_element_better(arr: list) -> tuple:
    
    largest= float('-inf')
    sec_largest= float('-inf')
    
    for i in range(len(arr)):
        if arr[i] > largest:
            largest= arr[i]
    
    for i in range(len(arr)):
        if (arr[i] > sec_largest) and (arr[i] != largest):
            sec_largest= arr[i]
    
    return largest, sec_largest


# function call
arr= [3,2,1,5,2,8]
largest, sec_largest= find_second_largest_element_better(arr)
print(largest, sec_largest)
```

```python
"""
Optimal Approach: 
1. Do everything in single pass
2. Intialize largest as 1st element, sec_largest as -inf
3. In 1st pass, find element greater than largest, so sec_largest becomes the largest and largest becomes new element: O(n)
4. Another condition being array element != largest and array element > sec_largest one, so update sec_largest

Time comlexity: O(n)
"""

def find_second_largest_element_optimal(arr: list) -> tuple:
    
    largest= arr[0]
    sec_largest= float('-inf')
    
    for i in range(len(arr)):
        if arr[i] > largest:
            sec_largest= largest
            largest= arr[i]
            
        elif arr[i] != largest and arr[i] > sec_largest:
            sec_largest= arr[i]
    
    return largest, sec_largest


# function call
arr= [3,2,1,5,2,8]
largest, sec_largest= find_second_largest_element_optimal(arr)
print(largest, sec_largest)
```
